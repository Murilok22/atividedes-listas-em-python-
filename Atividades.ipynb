```python
# Etapa 1 - mesngagem de boas vindas
print("====== Bem vindo ao sacolão da GENERATION ======\n")

# Etapa 2 - carregar dados - criar listas
produtos = ["Banana", "Melancia", "Jambolão"]
precos = [5,19,29.99]

# Etapa 3 - estrutura condicionada

# Etapa 3.1 - vamos mostrar os produtos
for i in range(len(produtos)):
    print(f"{i+1} - {produtos[i]} - R$ {precos[i]}")

print("\n")

for produto, preco in zip(produtos, precos):
    print(f"O produto {produto}, custa R$ {preco}")

print("\n")

resposta = input("D0eseja cadastar um novo produto? (s/n) ")
if resposta.lower() == "s":
    novo_produto = input("Digite o nome do novo produto: ")
    novo_preco = float(input("Digite o preço do novo produto: "))
    #anexando novo produto
    produtos.append(novo_produto)
    precos.append(novo_preco)
    print("\n")
    for i,j in zip(produtos, precos):
     print(f"O produto {i}, custa R$ {j}")
```
